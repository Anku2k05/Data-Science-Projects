```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data = {
    "Hours_Studied": [2,3, 4,5, 6,7, 8, 10],
    "Attendance": [60,64, 70, 75, 75,80, 85, 95],
    "Previous_Marks": [50,53, 55, 58,65,69, 75, 85],
    "Marks": [45,50, 55,60, 65,70, 78, 90]
}
df=pd.DataFrame(data)
```

```python
X = df[["Hours_Studied", "Attendance", "Marks"]]
```

```python
y = df["Marks"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=2
)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print(y_test)
print(y_pred)
```

```python
Hours_Studied = 6
Attendance = 70
Previous_Marks = 80

Pred_mark = model.predict([[Hours_Studied, Attendance, Previous_Marks]])

print(Pred_mark)
```
