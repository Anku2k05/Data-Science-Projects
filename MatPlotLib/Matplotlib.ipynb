Question 5
Highlight the highest and lowest sales using annotate() with arrows and save the figure as monthly_sales.png.



```python
import matplotlib.pyplot as plt
Month= ["Jan","Feb","Mar","Apr","May","Jun",
        "Jul","Aug","Sep","Oct","Nov","Dec"]
Sales= [4200,4800,5100,6200,6800,7200,
        7600,7300,6900,7500,8100,9200]
plt.figure(figsize=(12,6))
plt.plot(Month, Sales, color="darkblue", marker="o" , linewidth=3, markersize=8)
plt.title("Monthly Sales")
plt.xlabel("Months")
highest=Sales.index(max(Sales))
Lowest=Sales.index(min(Sales))
plt.annotate(text="Highest sale", xy=(Month[highest], Sales[highest]),xytext=("Nov", 9500), arrowprops=dict(arrowstyle="->", color="red", lw=2))
plt.annotate(text="Lowest sale", xy=(Month[Lowest], Sales[Lowest]),xytext=("Jan", 5000), arrowprops=dict(arrowstyle="->", color="red", lw=2))
plt.grid(linestyle="--", alpha=0.8)
plt.savefig("Monthly_sales.png")
plt.show()

```

Question 19
Display the mean and median using vertical lines and include their values in the title.


```python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(10)
salary=np.random.normal(55000,12000,400)
plt.hist(salary, bins=25, density=True)
mean_salary = np.mean(salary)
median_salary = np.median(salary)
plt.axvline(mean_salary, color="red",linestyle="--",linewidth=2, label="mean salary")
plt.axvline(median_salary, color="Black", linestyle="-.",linewidth=2, label="median salary")
plt.title(f"Mean = {mean_salary}, Median = {median_salary}")
plt.show()
```

Question 30 – Mini Project
Using the same dataset, create a professional 3 × 2 Business Analytics Dashboard containing:
Monthly Sales Line Chart with value annotations.
Grouped Bar Chart comparing Sales, Profit, and Expenses.
Pie Chart showing annual Expenses by quarter.
Histogram of Orders.
Scatter Plot of Customers vs Sales (marker size = Orders, marker color = Profit).
Area Chart showing cumulative Profit using fill_between().


```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun","Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Sales": [4200, 4800, 5100, 6200, 6800, 7200,7600, 7300, 6900, 7500, 8100, 9200],
    "Profit": [850, 980, 1100, 1350, 1520, 1680,1810, 1760, 1650, 1820, 1980, 2250],
    "Expenses": [3350, 3820, 4000, 4850, 5280, 5520,5790, 5540, 5250, 5680, 6120, 6950],
    "Customers": [210, 235, 248, 275, 301, 320,345, 338, 326, 349, 371, 405],
    "Orders": [180, 195, 205, 225, 245, 262,278, 270, 258, 281, 296, 325],
    "Returns": [8, 10, 9, 12, 11, 13,15, 14, 12, 11, 10, 9]
}
df = pd.DataFrame(data)
plt.figure(figsize=(14,12))
plt.subplot(3,2,1)
plt.plot(df["Month"], df["Sales"])
for Month, Sale in zip(df["Month"], df["Sales"]):
    plt.text(Month, Sale + 200, Sale, ha="center")
plt.title("Monthly Sales bar chart")

plt.subplot(3,2,2)
x = range(len(df))
plt.bar([i-0.2 for i in x], df["Sales"], width=0.2, label="Sales")
plt.bar(x, df["Profit"], width=0.2, label="Profit")
plt.bar([i+0.2 for i in x], df["Expenses"], width=0.2, label="Expenses")

plt.subplot(3,2,3)
quarter_expense = [
    sum(df["Expenses"][0:3]),
    sum(df["Expenses"][3:6]),
    sum(df["Expenses"][6:9]),
    sum(df["Expenses"][9:12])
]
plt.pie(quarter_expense,labels=["Q1","Q2","Q3","Q4"],autopct="%1.1f%%")
plt.title("Expenses by Quarter")

plt.subplot(3,2,4)
plt.hist(df["Orders"], bins=5)
plt.title("Orders Distribution")

plt.subplot(3,2,5)
plt.scatter(df["Customers"],df["Sales"],s=df["Orders"],c=df["Profit"],cmap="viridis")
plt.colorbar(label="Profit")
plt.title("Scatterplot: Customers vs Sales")

plt.subplot(3,2,6)
plt.fill_between(df["Month"], df["Profit"],color="skyblue",alpha=0.7)
plt.title("Profit Area Chart")

plt.savefig("business_analytics_dashboard.png")
plt.show()
```
